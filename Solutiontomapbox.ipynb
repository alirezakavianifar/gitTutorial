```python

import plotly.graph_objects as go

# Sample data with latitudes, longitudes, and labels
latitudes = [37.7749, 34.0522, 40.7128, 36.1627, 39.7392]  # Added more cities
longitudes = [-122.4194, -118.2437, -74.0060, -86.7816, -104.9903]
cities = ['San Francisco', 'Los Angeles', 'New York', 'Nashville', 'Denver']
colors = ['gold', 'blue', 'red', 'green', 'purple']  # Different colors for each city

# Define heads and subordinates
heads = ['San Francisco', 'New York']
subordinates = ['Los Angeles', 'Nashville', 'Denver']

# Your Mapbox access token
mapbox_token = "pk.eyJ1IjoidGF2YWtvbGltdG4iLCJhIjoiY20wdjU0NWV3MTh1aDJqcHdrYjRqZTVtYyJ9.9Oe7458AhdlS5trfzOKnJw"

# Create an empty figure
fig = go.Figure()

# Loop through each city and add it as a separate trace for the map
for lat, lon, city, color in zip(latitudes, longitudes, cities, colors):
    if city in heads:
        # Use star symbol for heads
        marker_symbol = 'star'
        marker_size = 15
    else:
        # Use circle symbol for subordinates
        marker_symbol = 'circle'
        marker_size = 10

    # Add scattermapbox trace for the map
    fig.add_trace(go.Scattermapbox(
        lat=[lat],
        lon=[lon],
        mode='markers',
        marker=go.scattermapbox.Marker(
            size=marker_size,
            symbol=marker_symbol,
            color=color
        ),
        text=[city],  # Display city name on hover
        name=city,  # Set the legend label to the city name
        showlegend=False  # Hide the legend for map trace
    ))

# Add separate scatter traces just for the legend
for city, color in zip(cities, colors):
    if city in heads:
        # Star symbol for heads
        fig.add_trace(go.Scatter(
            x=[None], y=[None],  # Invisible points
            mode='markers',
            marker=dict(
                size=15,
                symbol='star',
                color=color
            ),
            name=city,  # Legend entry
            showlegend=True  # Display legend
        ))
    else:
        # Circle symbol for subordinates
        fig.add_trace(go.Scatter(
            x=[None], y=[None],  # Invisible points
            mode='markers',
            marker=dict(
                size=10,
                symbol='circle',
                color=color
            ),
            name=city,  # Legend entry
            showlegend=True  # Display legend
        ))

# Set the layout, including mapbox style and zoom
fig.update_layout(
    mapbox=dict(
        style="streets",
        zoom=3,
        center=dict(lat=37.7749, lon=-95.4194),
        accesstoken=mapbox_token
    ),
    height=500,
    margin=dict(l=0, r=0, t=0, b=0),

    # Set the legend to be on the right
    legend=dict(
        x=1,
        y=0.5,
        traceorder="normal",
        font=dict(
            size=12,
            color="black"
        ),
        bgcolor="LightGray",
        bordercolor="Black",
        borderwidth=1
    )
)

# Show the figure
fig.show()
```
